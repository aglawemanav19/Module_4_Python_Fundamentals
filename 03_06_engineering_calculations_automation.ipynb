## 2. Worked example: heat-duty estimate (15 min)

For a liquid stream, approximate heat duty as $Q = \dot m C_p \Delta T$. If mass flow is in kg/h and $C_p$ is kJ/(kg·°C), the result is kJ/h. Dividing by 3600 gives kW.

```python
def heat_duty_kw(mass_flow_kg_h, cp_kj_kg_c, inlet_c, outlet_c):
    """Estimate sensible heat duty in kW; positive means heating."""
    delta_t_c = outlet_c - inlet_c
    return mass_flow_kg_h * cp_kj_kg_c * delta_t_c / 3600

duty_kw = heat_duty_kw(1200, 4.18, 25, 70)
print(f'Estimated heat duty = {duty_kw:.2f} kW')
```

**Try it (5 min):** Calculate the duty for a different flow rate and outlet temperature. Predict whether it will increase or decrease before running.

**Challenge:** Add a parameter `efficiency` and return the required utility duty when efficiency is below 1.

```python
def heat_duty_kw(mass_flow_kg_h, cp_kj_kg_c, inlet_c, outlet_c, efficiency=1.0):
    """Estimate required utility duty in kW; positive means heating."""
    if not 0 < efficiency <= 1:
        raise ValueError('efficiency must be greater than 0 and no greater than 1')

    delta_t_c = outlet_c - inlet_c
    process_duty_kw = mass_flow_kg_h * cp_kj_kg_c * delta_t_c / 3600
    return process_duty_kw / efficiency

# Original case, with no utility loss.
duty_kw = heat_duty_kw(1200, 4.18, 25, 70)
print(f'Estimated heat duty = {duty_kw:.2f} kW')

# Different flow rate and outlet temperature: both increase the heating duty.
trial_duty_kw = heat_duty_kw(1500, 4.18, 25, 80)
print(f'Trial heat duty = {trial_duty_kw:.2f} kW')
print('Prediction: the trial duty will increase.')

# At 80% efficiency, more utility is required than the process duty.
efficient_duty_kw = heat_duty_kw(1500, 4.18, 25, 80, efficiency=0.80)
print(f'Required utility duty at 80% efficiency = {efficient_duty_kw:.2f} kW')
```

## 3. A batch-quality report (20 min)

The script below inspects several conversion values. Notice how each building block has a role:

- a **list** holds batch data;
- a **function** creates a consistent status;
- a **loop** checks every batch;
- a **conditional** identifies batches needing attention;
- a counter summarises the result.

```python
conversions_percent = [91.2, 87.5, 93.0, 84.1, 89.8]
target_percent = 90.0

def quality_status(conversion_percent, target_percent):
    if conversion_percent >= target_percent:
        return 'PASS'
    return 'REVIEW'

review_count = 0
for batch_number, conversion_percent in enumerate(conversions_percent, start=1):
    status = quality_status(conversion_percent, target_percent)
    print(f'Batch {batch_number}: {conversion_percent:.1f}% → {status}')
    if status == 'REVIEW':
        review_count = review_count + 1

print(f'\nSummary: {review_count} of {len(conversions_percent)} batches require review.')
```

**Run, then modify (5 min):** Change one conversion to 90.0. Does it pass? Why? Add a sixth batch.

**Challenge:** Calculate and print the average conversion. Add `EXCELLENT` for conversion ≥ 95%.

```python
conversions_percent = [91.2, 87.5, 93.0, 84.1, 89.8, 90.0, 96.5]
target_percent = 90.0
excellent_percent = 95.0

def quality_status(conversion_percent, target_percent):
    if conversion_percent >= excellent_percent:
        return 'EXCELLENT'
    if conversion_percent >= target_percent:
        return 'PASS'
    return 'REVIEW'

review_count = 0
for batch_number, conversion_percent in enumerate(conversions_percent, start=1):
    status = quality_status(conversion_percent, target_percent)
    print(f'Batch {batch_number}: {conversion_percent:.1f}% → {status}')
    if status == 'REVIEW':
        review_count = review_count + 1

average_conversion = sum(conversions_percent) / len(conversions_percent)
print(f'\nSummary: {review_count} of {len(conversions_percent)} batches require review.')
print(f'Average conversion: {average_conversion:.1f}%')
```

## 4. Build your own mini automation script (15 min)

Choose one scenario below. Work in pairs, then test at least three cases (including a boundary).

**Option A — Tank monitor:** Given a list of level percentages, report `LOW` below 20%, `HIGH` above 85%, otherwise `OK`; print totals for each status.

**Option B — Pump energy estimate:** Given flow rates for several hours, calculate total pumped volume. If a flow is negative, flag it as invalid rather than including it.

**Option C — Heat exchanger checker:** For several outlet temperatures, calculate heat duty and label values over a chosen limit `CHECK UTILITY`.

Use clear variable names and document units in a Markdown cell or code comments.

```python

# Tank levels in percent (%).
level_percentages = [15.0, 20.0, 42.5, 85.0, 92.0, 18.5]


def tank_status(level_percent):
    if level_percent < 20:
        return 'LOW'
    if level_percent > 85:
        return 'HIGH'
    return 'OK'


status_totals = {'LOW': 0, 'OK': 0, 'HIGH': 0}
for reading_number, level_percent in enumerate(level_percentages, start=1):
    status = tank_status(level_percent)
    status_totals[status] += 1
    print(f'Reading {reading_number}: {level_percent:.1f}% - {status}')

print('\nStatus totals:')
for status, total in status_totals.items():
    print(f'{status}: {total}')
```
